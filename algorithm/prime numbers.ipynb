# 求小于某个数的所有素数
要求：在terminal中输入一个数，求小于等于这个数的 ***所有素数*** 集合，并按照每行8个进行输出

```python
def is_prime(x):
    if x < 2:
        return False
    if x == 2:
        return True
    if x % 2 == 0:
        return False
    for y in range(3, int(x**0.5) + 1, 2):
        if x % y == 0:
            return False
    return True

# 输入验证改进
while True:
    try:
        m = int(input("请输入一个正整数: "))
        if m > 0: break
        print("必须输入正整数！")
    except ValueError:
        print("输入内容必须是数字！")

primes = []
for x in range(2, m):
    if is_prime(x):
        primes.append(x)

# 空结果处理
if not primes:
    print(f"没有小于 {m} 的素数")
    exit()

# 输出格式优化
print(f"小于 {m} 的素数包括：")
for idx, num in enumerate(primes, 1):
    end_char = "\n" if idx % 8 == 0 else " "
    print(f"{num:>5d}", end=end_char)
if len(primes) % 8 != 0:  # 最后换行修正
    print()
```
