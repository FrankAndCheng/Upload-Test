**算法 1：** 为解决 8 皇后问题，初步想法为，以随机数挑选皇后棋子的位置，一层一层判断是否符合要求


```python
import random

def is_safe(arr_ran, row, col):
    """Check if placing a queen at (row, col) is safe"""
    for i in range(row):
        # Check column and diagonal conflicts
        if (arr_ran[i] == col or
            abs(arr_ran[i] - col) == abs(i - row)):
            return False
    return True

def solve_nqueens_random():
    """Solve N-Queens with random selection and backtracking"""
    print("A possible solution is:")

    arr_ran = []
    available_cols = list(range(8))

    def backtrack(row):
        if row == 8:
            return True

        # Shuffle available columns for randomness
        random.shuffle(available_cols)

        for col in available_cols[:]:
            if is_safe(arr_ran, row, col):
                arr_ran.append(col)
                available_cols.remove(col)

                if backtrack(row + 1):
                    return True

                # Backtrack
                arr_ran.pop()
                available_cols.append(col)

        return False

    if backtrack(0):
        # Display the solution
        for row in range(8):
            queen_col = arr_ran[row]
            for col in range(8):
                if col == queen_col:
                    print("|Q", end="")
                else:
                    print("| ", end="")
            print("|")
        return arr_ran
    else:
        print("No solution found")
        return None

# Alternative: Fixed version closer to original structure
def solve_nqueens_original_style():
    """Fixed version maintaining original code structure"""
    print("A possible solution is:")

    max_attempts = 1000  # Prevent infinite loops
    attempts = 0

    while attempts < max_attempts:
        attempts += 1
        arr_ran = []
        arr_change = list(range(8))
        success = True

        for m in range(8):
            if not arr_change:  # No more columns available
                success = False
                break

            placed = False
            tries = 0

            while not placed and tries < len(arr_change):
                tries += 1
                n = random.randint(0, len(arr_change) - 1)
                col = arr_change[n]

                # Check if this position is safe
                safe = True
                for j in range(m):
                    if abs(col - arr_ran[j]) == abs(m - j):
                        safe = False
                        break

                if safe:
                    arr_ran.append(col)
                    arr_change.remove(col)
                    placed = True

            if not placed:
                success = False
                break

        if success:
            # Display the solution
            for row in range(8):
                queen_col = arr_ran[row]
                for col in range(8):
                    if col == queen_col:
                        print("|Q", end="")
                    else:
                        print("| ", end="")
                print("|")
            return arr_ran

    print("No solution found after", max_attempts, "attempts")
    return None

# Run the corrected version
result = solve_nqueens_random()
print(f"\nSolution: {result}")
```

**算法 2：**

- 随机生成一个数列，直接判断其数列符合要求与否
- 倘若数列不符合要求，则推翻此数列，再重新随机生成


```python
import random

print("A possible solution is:")
Flag = False
while not Flag:
    ran = range(0, 8)
    arr_change = list(ran)
    arr_ran = list()
    for m in range(0, 8):
        n = random.randint(0, 7 - m)
        arr_ran.append(arr_change[n])
        arr_change.remove(arr_change[n])
    for i in range(1, 8):
        for j in range(0, i):
            if abs(arr_ran[i] - arr_ran[j]) == (i - j):
                Flag = False
                break
            else:
                Flag = True
        if not Flag:
            break

print("最终满足要求的位置为", arr_ran)

for i in range(0, 8):
    for j in range(0, 8):
        if j == arr_ran[i]:
            print("|Q", end="")
        else:
            print("| ", end="")
    print("|")
```

**算法3**:
    标准解决方法

```python
def conflict(state, nextColumn):
    nextRow = rows = len(state)
    for row in range(rows):
        column = state[row]
        if abs(column - nextColumn) in (0, nextRow - row):
            """
            如果差值等于0,两个皇后在同一列,则代表冲突,返回True;
            如果列的差值等与行的差,两个皇后在对角线上,则代表冲突,返回True;
            """
            return True
    return False


def queens(num, state=()):
    """
    采用生成器的方式来产生每一个皇后的位置,并用递归来实现下一个皇后的位置
    num: 皇后的数量
    state: 标记已经排好的每个皇后的位置
    """
    for pos in range(num):  # 八皇后的数量N=0, 1, 2, 3, 4, 5, 6 , 7 你要在哪一列放置皇后
        # 如果不冲突，则递归构造棋盘。
        if not conflict(state, pos):  # 回溯法的体现
            # 如果棋盘状态state已经等于num-1，即到达倒数第二行，而这时最后一行皇后又没冲突，直接yield，打出其位置(pos, )
            if len(state) == num - 1:  # state=()
                yield (pos,)
            else:  # (0, )
                for result in queens(num, state + (pos,)):
                    yield (pos,) + result


def prettyprint(solution):

    def line(pos, length=len(solution)):
        return "| " * pos + "|Q" + "| " * (length - pos - 1) + "|"

    for pos in solution:
        print(line(pos))


if __name__ == "__main__":
    solutions = queens(8)
    for index, solution in enumerate(solutions):
        print("第%d种解决方案:" % (index + 1), solution)
        prettyprint(solution)
        print("\n" + "*" * 50 + "\n")
```
